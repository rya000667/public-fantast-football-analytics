## Load Data

```python
# File Paths
from os import path


BASE_DIR = Path.cwd()
DATA_DIR = BASE_DIR / "data/college_data"

# Hold DataFrames
dfs = {}

# Read in all JSON files in the data directory and normalize them into DataFrames
for item in DATA_DIR.iterdir():
    if item.is_file():
        with open(item, encoding="utf-8") as f:
            dfs[item.stem] = pd.json_normalize(json.load(f), sep="_")
        print(f"Loaded {item.stem} into DataFrame with shape {dfs[item.stem].shape}")
```

## Prune Unnecessary Features

Fantasy Football is primarily focused on offensive statistics, the following code identifies irrelevant categories and drops them before creating a player level aggregate. 

### Prune Player Stats

```python

# Make sure all stat categories are the same between 2024 and 2025 stats
# Save the stat categories that are in both 2024 and 2025 stats to a variable for later use

stat_categories = set(dfs['2024_stats'].category.unique()) - set(dfs['2025_stats'].category.unique())
if len(stat_categories) > 0:
    print(f"Stat categories in 2024 but not in 2025: {stat_categories}")

else: 
    print("All stat categories are the same between 2024 and 2025 stats.")
    stat_categories = set(dfs['2025_stats'].category.unique()) & set(dfs['2024_stats'].category.unique())
    print(f"Stat categories in both 2024 and 2025: {stat_categories}")

all_stats =  pd.concat([dfs["2024_stats"], dfs["2025_stats"]], ignore_index=True)
display(all_stats.info())
```

```python
# Retained Features 
stats_retain_features = ['receiving', 'passing', 'rushing', 'fumbles', 'kicking']

# Filter DataFrame to only include rows with the retained features
all_stats_filtered = all_stats[all_stats['category'].isin(stats_retain_features)]
print(f"Features Removed: {set(all_stats['category'].unique()) - set(all_stats_filtered['category'].unique())}")
```

### Prune Player Draft Data

```python
draft_features = set(dfs['2025_draft'].columns) - set(dfs['2026_draft'].columns)
if len(draft_features) > 0:
    print(f"Draft features in 2025 but not in 2026: {draft_features}")

else: 
    print("All draft features are the same between 2025 and 2026 stats.")
    draft_features = set(dfs['2025_draft'].columns) & set(dfs['2026_draft'].columns)
    print(f"Draft features in both 2025 and 2026: {draft_features}")

all_draft = pd.concat([dfs["2025_draft"], dfs["2026_draft"]], ignore_index=True)
display(all_draft.info())

# View Features in the all_draft DataFrame
print(all_draft.columns)
```

```python
all_draft
```

```python
draft_retain_features = ['collegeAthleteId', 'nflAthleteId', 'collegeId', 'collegeTeam',
       'collegeConference', 'nflTeamId', 'nflTeam', 'year', 'overall', 'round',
       'pick', 'name', 'position', 'height', 'weight', 'preDraftRanking',
       'preDraftPositionRanking', 'preDraftGrade', 'hometownInfo_city',
       'hometownInfo_state']

# Filter DataFrame to only include rows with the retained features
all_draft_filtered = all_draft[draft_retain_features]
print(f"Features Removed: {set(all_draft.columns) - set(all_draft_filtered.columns)}")
```
